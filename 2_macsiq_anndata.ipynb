# Creating the AnnData File

#  Import required packages

```python
import pandas as pd
import numpy as np
import anndata as ad
import scanpy as sc
from pathlib import Path
import os
import yaml
%matplotlib inline
%config InlineBackend.figure_format='retina'
```

# Required Inputs

```python
with open("config.yaml", "r") as f:
    config = yaml.safe_load(f)

# Set up directories
metadata_dir = Path(config['metadata_dir'])
preprocessed_dir = Path(config['preprocessed_dir'])
results_dir = Path(config['results_dir'])
# make a roi list of directories in preprocessed_dir that are not .DS_Store
roi_list = [roi for roi in preprocessed_dir.iterdir() if roi.is_dir() and roi.name != '.DS_Store']

patients_metadata = pd.read_csv(metadata_dir / 'new_patients_metadata.csv')
```

# Add the Patients Metadata to the ROIs

```python
for roi in roi_list:
    exp = pd.read_csv(roi / "exp.csv")
    meta = pd.read_csv(roi / "metadata.csv")
    if "exp_name" in meta.columns:
        print("Metadata already has exp_name column")
        continue
    else:
        pass
    print("Adding exp_name column to metadata")
    meta["exp_name"] = roi.name
    meta = meta.merge(patients_metadata, on="exp_name")
    meta.to_csv(roi / "metadata.csv", index=False)
```

## Create the Anndata Object for each ROI

```python
for roi in roi_list:
    if os.path.exists(roi / "adata.h5ad"):
        print("adata file already exists")
        continue
    else:
        pass
    print("Creating adata file")
    exp = pd.read_csv(roi / "exp.csv")
    meta = pd.read_csv(roi / "metadata.csv")
    #get the cells coordinates
    cells_coordinates = np.array([meta["Cell Center X"], meta["Cell Center Y"]]).T
    adata = ad.AnnData(X=exp.iloc[:,1:], obs=meta, obsm={"spatial": cells_coordinates})
    #write adata file
    adata.write(roi / "adata.h5ad")
```

# Creating a concatenated adata object

```python
adata_list = []
for i,roi in enumerate(roi_list):
    adata = ad.read_h5ad(roi / "adata.h5ad")
    adata.obs_names = [f"{i}_{j}" for j in range(adata.n_obs)]
    adata_list.append(adata)
adata = ad.concat(adata_list, join="outer", axis=0)
adata.obs.cell_type = adata.obs.cell_type.astype("category")
adata.obs.run = adata.obs.run.astype("category")
adata.obs.patient_ID = adata.obs.patient_ID.astype("category")
adata.obs.exp_name = adata.obs.exp_name.astype("category")

adata.obs["patient_exp"] = adata.obs["patient_ID"].astype(str) + "_" + adata.obs["rois"].astype(str)
extra_patient_info = pd.read_excel(metadata_dir / 'extra_patients_info.xlsx', sheet_name='Detailed info')
extra_patient_info.rename(columns={'Patient ID': 'patient_ID'}, inplace=True)#get the survival time in months and event
extra_patient_info["survival_time_months"] = extra_patient_info["Survival after NMC diagnosis"].str.split(" month").str[0].astype(float)
adata.obs = adata.obs.merge(extra_patient_info, how='left', left_on='patient_ID', right_on='patient_ID')
adata.obs["patient_exp"] = adata.obs["patient_ID"].astype(str) + "_ROI" + adata.obs["rois"].astype(str)
adata.obs["site of biopsy/resection for MACSima sample"].unique()
#fix some names of the sites
adata.obs["site of biopsy/resection for MACSima sample"] = adata.obs["site of biopsy/resection for MACSima sample"].replace({
    "subcutaneous metastasis (thigh left)": "Metastatic left thigh",
    "retroportal lymph node (mainly tumor)": "retroportal lymph node",
})

adata.raw = adata.copy()
adata.write(results_dir / "adata.h5ad")
```

```python
#conctenate all the adata files
adata_list = []
for roi in roi_list:
    adata = ad.read_h5ad(roi / "adata.h5ad")
    if adata.X.shape[1] != 65:
        continue
    adata_list.append(adata)
adata = ad.concat(adata_list)
#make obs names unique
adata.obs_names_make_unique()
#make the categorical variables
adata.obs.cell_type = adata.obs.cell_type.astype("category")
adata.obs.run = adata.obs.run.astype("category")
adata.obs.patient_ID = adata.obs.patient_ID.astype("category")
adata.obs.exp_name = adata.obs.exp_name.astype("category")
#save the raw adata.X
adata.raw = adata.copy()
#write it down as a h5ad file
adata.write(results_dir / "adata.h5ad")
```
