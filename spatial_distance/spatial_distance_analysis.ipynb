# Loading data

```python
adata = ad.read_h5ad(os.path.join(data_dir, "adata.h5ad"))
```

# Define tumor contours

```python
adata = tumor_mask_and_distance_to_contour(adata, 
                                           plot=True, 
                                           show_legend=False,
                                            grid_size=30,
                                           show_titles=False, 
                                           save_svg = True, 
                                           svg_dir = os.path.join(results_dir, "contours"), 
                                           dpi=300)
```

```python
#Add distance layers..
adata = add_tumor_layers(
    adata,
    dist_col="tumor_signed_dist_to_contour_um",  # make sure this column exists
    step=25,                                     # 25 µm (or pixels if still in px)
    out_col="tumor_layer"
)
```

# Save the distances of the each cell to the tumor contour 

```python
#Let's export it 
distances_df = adata.obs.loc[:, ['patient_exp', "patient_ID", "cell_category", "Cell Center X", "Cell Center Y", 'tumor_signed_dist_to_contour_um', 'tumor_dist_to_contour_um', "tumor_layer", "tumor_total_area_um2", "non_tumor_area_um2"]].reset_index(drop=True)
distances_df.to_csv(os.path.join(data_dir, "cell_distances_to_tumor_contour.csv"), index=False)
```

# Calculate distances to tumor cells in the contour

```python
adata.obs["cell_category_contour"] = adata.obs.apply(
    lambda row: "Tumor cells (contour)" if row["tumor_layer"] in ["stroma 0-25 µm", "tumor 0-25 µm", "tumor 25+ µm"] 
    else row["cell_category"], axis=1)
```

```python
sm.tl.spatial_distance(adata, x_coordinate="Cell Center X", y_coordinate="Cell Center Y", phenotype="cell_category_contour", imageid="patient_exp")
```

```python
distances_scimap = adata.uns["spatial_distance"].copy()
distances_scimap["cell_category"] = adata.obs["cell_category"].values
#replace patient_exp values from NMC_ to NUT_
distances_scimap["patient_exp"] = adata.obs["patient_exp"].str.replace("NMC_", "NUT_", regex=False).values
distances_scimap["patient_ID"] = adata.obs["patient_ID"].str.replace("NMC_", "NUT_", regex=False).values
distances_scimap = distances_scimap[["patient_exp", "patient_ID", "cell_category", "Tumor cells (contour)"]]
distances_scimap["Tumor cells (contour)"] = distances_scimap["Tumor cells (contour)"].astype(float) * 0.17
#rename it to Tumor cells
distances_scimap.rename(columns={"Tumor cells (contour)": "Tumor cells"}, inplace=True)
#save the distances_scimap dataframe to csv
distances_scimap.to_csv(os.path.join(data_dir, "distances_to_tumor_cells_scimap.csv"), index=False)
```

# Calculate enrichment per layer

```python
enrich_df = compute_patient_layer_enrichment(adata)
enrich_df
```

```python
#calculate median layer_proportion, global_proportion, fold_enrichment, log2_fold_enrichment for each cell type in each layer across patients
enrich_df = enrich_df.groupby(["patient_ID", "tumor_layer", "cell_type"]).agg({
    "layer_proportion": "median",
    "global_proportion": "median",
    "fold_enrichment": "median",
    "log2_fold_enrichment": "median"
}).reset_index()
enrich_df["diff"] = enrich_df["layer_proportion"] - enrich_df["global_proportion"]
enrich_df 
```

```python
order = ["distal_stroma", "stroma_75_100", "stroma_50_75", "stroma_25_50", "stroma_tumor", "tumor_0_25", "tumor_25+"]
enrich_df["tumor_layer"] = pd.Categorical(
    enrich_df["tumor_layer"], categories=order, ordered=True
)
enrich_df
```

```python
enrich_df.to_csv(os.path.join(data_dir, "layer_enrichment_results.csv"), index=False)
```
