```python
import os
import napari
import dask.array as da
from dask_image.imread import imread
```

```python
# ── 1. Image folder & file discovery ──────────────────────────────────────────
image_folder = "/Volumes/T7/Run145/ROI10/"
images = sorted(os.listdir(image_folder))

def find(keyword, index=0):
    matches = [img for img in images if keyword in img]
    if not matches:
        raise FileNotFoundError(f"No image found for keyword: {keyword}")
    return os.path.join(image_folder, matches[index])

# ── 2. Lazy load ───────────────────────────────────────────────────────────────
def lazy_load(filepath):
    arr = imread(filepath)
    if arr.ndim == 3:
        arr = arr[0]
    return arr

# ── 3. Channel definitions ─────────────────────────────────────────────────────
channels = [
    ("DAPI",       "DAPI",            (0, 4000), 1, "translucent", 0.5,  True ),
    ("CD20",       "CD20Cytoplasmic", (0, 3000), 0, "additive",    1.0,  False),
    ("CD4",        "CD4_",            (0, 3000), 0, "additive",    1.0,  False),
    ("CD8a",       "CD8a_",           (0, 3000), 0, "additive",    1.0,  False),
    ("MRP14",      "MRP14",           (0, 3000), 0, "additive",    1.0,  False),
    ("Arginase-1", "Argi",            (0, 3000), 0, "additive",    1.0,  False),
    ("CD15",       "CD15_",           (0, 3000), 0, "additive",    1.0,  False),
    ("CD204",      "CD204",           (0, 3000), 0, "additive",    1.0,  False),
    ("PRAME",      "PRAME",           (0, 3000), 0, "additive",    1.0,  False),
]

# ── 4. Open napari ─────────────────────────────────────────────────────────────
viewer = napari.Viewer(title="ROI17 Multiplex IF")

for name, keyword, clim, match_index, blending, opacity, visible in channels:
    path = find(keyword, index=match_index)
    lazy_arr = lazy_load(path)
    viewer.add_image(
        lazy_arr,
        name=name,
        contrast_limits=clim,
        blending=blending,
        opacity=opacity,
        visible=visible,
    )

napari.run()
```
