# Creating the AnnData File

#  Import required packages

```python
import pandas as pd
import numpy as np
import anndata as ad
import scanpy as sc
from pathlib import Path
import os
import yaml
%matplotlib inline
%config InlineBackend.figure_format='retina'
```

# Required Inputs

```python
with open("../config.yaml", "r") as f:
    config = yaml.safe_load(f)

# Set up directories
metadata_dir = Path(config['metadata_dir'])
preprocessed_dir = Path(config['preprocessed_dir'])
results_dir = Path(config['results_dir'])
# make a roi list of directories in preprocessed_dir that are not .DS_Store
roi_list = [roi for roi in preprocessed_dir.iterdir() if roi.is_dir() and roi.name != '.DS_Store']

patients_metadata = pd.read_csv(metadata_dir / 'new_patients_metadata.csv')
```

# Add the Patients Metadata to the ROIs

```python
for roi in roi_list:
    exp = pd.read_csv(roi / "exp.csv")
    meta = pd.read_csv(roi / "metadata.csv")
    if "exp_name" in meta.columns:
        print("Metadata already has exp_name column")
        continue
    else:
        pass
    print("Adding exp_name column to metadata")
    meta["exp_name"] = roi.name
    meta = meta.merge(patients_metadata, on="exp_name")
    meta.to_csv(roi / "metadata.csv", index=False)
```

## Create the Anndata Object for each ROI

```python
for roi in roi_list:
    if os.path.exists(roi / "adata.h5ad"):
        print("adata file already exists")
        continue
    else:
        pass
    print("Creating adata file")
    exp = pd.read_csv(roi / "exp.csv")
    meta = pd.read_csv(roi / "metadata.csv")
    #get the cells coordinates
    cells_coordinates = np.array([meta["Cell Center X"], meta["Cell Center Y"]]).T
    adata = ad.AnnData(X=exp.iloc[:,1:], obs=meta, obsm={"spatial": cells_coordinates})
    #write adata file
    adata.write(roi / "adata.h5ad")
```

# Creating a concatenated adata object

```python
adata_list = []
for i,roi in enumerate(roi_list):
    adata = ad.read_h5ad(roi / "adata.h5ad")
    adata.obs_names = [f"{i}_{j}" for j in range(adata.n_obs)]
    adata_list.append(adata)
adata = ad.concat(adata_list, join="outer", axis=0)
adata.obs.cell_type = adata.obs.cell_type.astype("category")
adata.obs.run = adata.obs.run.astype("category")
adata.obs.patient_ID = adata.obs.patient_ID.astype("category")
adata.obs.exp_name = adata.obs.exp_name.astype("category")

adata.obs["patient_exp"] = adata.obs["patient_ID"].astype(str) + "_" + adata.obs["rois"].astype(str)
extra_patient_info = pd.read_excel(metadata_dir / 'extra_patients_info.xlsx', sheet_name='Detailed info')
extra_patient_info.rename(columns={'Patient ID': 'patient_ID'}, inplace=True)#get the survival time in months and event
extra_patient_info["survival_time_months"] = extra_patient_info["Survival after diagnosis of malignacy"].str.extract(r"(\d+)\s*[mM]onths").astype(float)
adata.obs = adata.obs.merge(extra_patient_info, how='left', left_on='patient_ID', right_on='patient_ID')
adata.obs["patient_exp"] = adata.obs["patient_ID"].astype(str) + "_ROI" + adata.obs["rois"].astype(str)
adata.obs["site of biopsy/resection for MACSima sample"].unique()
#fix some names of the sites
adata.obs["site of biopsy/resection for MACSima sample"] = adata.obs["site of biopsy/resection for MACSima sample"].replace({
    "subcutaneous metastasis (thigh left)": "Metastatic left thigh",
    "retroportal lymph node (mainly tumor)": "retroportal lymph node",
})

adata.raw = adata.copy()
adata.obs.reset_index(drop=True, inplace=True)
# adata.write(results_dir / "adata.h5ad")
```

## Add proper tissue definition and cell categories

```python
tissue_map = {
    'NMC_1': 'Head and Neck',
    'NMC_2': 'Lung',
    'NMC_4': 'Lung',
    'NMC_5': 'Head and Neck',
    'NMC_6': 'Lung',
    'NMC_7': 'Lung',
    'NMC_8': 'Lung',
    'NMC_9': 'Lung',
    'NMC_12': 'Head and Neck',
    'NMC_13': 'Lung',
    'NMC_14': 'Abdomen',
    'NMC_15': 'Lung',
    'NMC_16': 'Head and Neck',
    'NMC_21': 'Lung',
}


#Get the cell types and colors...

#DCs #Actin+_cells #Macrophages #T_cells #NK_cells #Lymphatic_endothelial_cells #B_cells #Plasma_cells #Granulocytes #Mast_cells #NFC #MDSCs #Endothelial_cells #Tumor_cells #Epithelial_cells
conditions = [
    adata.obs['cell_type'] == 'CD4+_T_cells',
    adata.obs['cell_type'] == 'CD8+_T_cells',
    adata.obs['cell_type'] == 'Treg_T_cells',
    adata.obs['cell_type'] == 'NK_cells',
    adata.obs['cell_type'] == 'M1_macrophages',
    adata.obs['cell_type'] == 'M2_macrophages',
    adata.obs['cell_type'] == 'M1_M2_macrophages',
    adata.obs['cell_type'] == 'B_cells',
    adata.obs['cell_type'] == 'Plasma_cells',
    adata.obs['cell_type'] == 'DCs',
    adata.obs['cell_type'] == 'Granulocytes',
    adata.obs['cell_type'] == 'Mast_cells',
    adata.obs['cell_type'] == 'MDSCs',
    adata.obs['cell_type'] == 'Lymphatic_endothelial_cells',
    adata.obs['cell_type'] == 'Endothelial_cells',
    adata.obs['cell_type'] == 'Actin+_cells',
    adata.obs['cell_type'] == 'Tumor_cells',
    adata.obs['cell_type'] == 'Epithelial_cells',
    adata.obs['cell_type'] == 'NFC'
]

# Define the corresponding labels
labels = [
    'CD4+ T cells',
    'CD8+ T cells',
    'Treg T cells',
    'NK cells',
    'M1-like macrophages',
    'M2-like macrophages',
    'M1/M2-like macrophages',
    'B cells',
    'PCs',
    'DCs',
    'Granulocytes',
    'Mast cells',
    'MDSCs',
    'Lymphatic endothelial cells',
    'Endothelial cells',
    'Actin+ cells',
    'Tumor cells',
    'Epithelial cells',
    'NFC'
]


# 1. Add all metadata mappings first
adata.obs['tumor_site'] = adata.obs['patient_ID'].map(tissue_map)
adata.obs['cell_category'] = np.select(conditions, labels, default='Other')

# 2. Filter bad ROIs
mask = ~adata.obs['patient_exp'].isin(["NMC_12_ROI6", "NMC_21_ROI23", "NMC_6_ROI6"])
adata = adata[mask.values].copy()

# 3. Sort through adata not obs directly
adata = adata[adata.obs.sort_values('patient_ID').index].copy()

# 4. Add spatial AFTER all filtering/sorting
adata.obsm["spatial"] = adata.obs[["Cell Center X", "Cell Center Y"]].values

# 5. Rename at the end
adata.obs["patient_ID"] = adata.obs["patient_ID"].str.replace("NMC", "NUT")
adata.obs["patient_exp"] = adata.obs["patient_exp"].str.replace("NMC", "NUT")
```

```python
#write the adata file
adata.write(results_dir / "adata.h5ad")
```
